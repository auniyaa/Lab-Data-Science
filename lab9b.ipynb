```r
# Load dataset
data("ToothGrowth")

# View data
head(ToothGrowth)

# Convert categorical variables to numeric (if needed)
# dose is already numeric, supp is categorical
df <- ToothGrowth
df$supp <- as.numeric(factor(df$supp))

# Compute correlation matrix
cor_matrix <- cor(df)

# View correlation
print(cor_matrix)

# Install package (if not installed)
install.packages("corrplot")

# Load library
library(corrplot)

# Plot heatmap
corrplot(cor_matrix, method = "color", addCoef.col = "black", tl.col = "black")
```

```r
# Load dataset
data("mtcars")

df <- mtcars

# 1. Log Transformation
log_df <- log(df)

# 2. Standard Scaling (Z-score)
zscore_df <- scale(df)

# 3. Min-Max Scaling
minmax <- function(x) {
  (x - min(x)) / (max(x) - min(x))
}

minmax_df <- as.data.frame(lapply(df, minmax))

# View results
head(log_df)
head(zscore_df)
head(minmax_df)
```
