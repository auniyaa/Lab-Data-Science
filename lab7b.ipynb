```r
library(readr)
library(dplyr)
library(lubridate)
# 1. Count by City
city_counts <- uforeport %>%count(City, sort = TRUE)

print("Top 3 Cities:")
head(city_counts, 3)

colour_count <- uforeport %>%count( Colors.Reported ,sort=TRUE)%>%
  filter(!is.na(Colors.Reported))
print("Top 3 colours:")
head(colour_count,3)

colour_count <- uforeport %>%count( Shape.Reported ,sort=TRUE)%>%
  filter(!is.na(Shape.Reported))
print("Top 3 Shape:")
head( colour_count,3)

state_counts <- uforeport %>%count(State, sort = TRUE)

print("Top 3 States:")
head(state_counts, 3)

ufo_with_year <- uforeport %>%mutate(DateTime = mdy_hm(Time),
         Year = year(DateTime))

year_counts <- ufo_with_year %>%count(Year, sort = TRUE)

print("Top 3 Years with most sightings:")
head(year_counts, 3)

```

```r
top_state_name <- uforeport %>%
  count(State, sort = TRUE) %>%
  slice(1) %>%
  pull(State)

# 3. Create a list of the categories you want to analyze for that State
categories <- c("City", "Shape.Reported", "Colors.Reported")

# 4. Use lapply to run the same logic across all categories at once
# This excludes NAs automatically for each category
top_state_stats <- lapply(categories, function(cat) {
  uforeport %>%
    filter(State == top_state_name) %>% # Look only at the top state
    filter(!is.na(!!sym(cat))) %>%      # Remove missing values
    count(!!sym(cat), sort = TRUE) %>%  # Count and sort
    head(1)                             # Keep only the #1 result
})

# Assign names to the list for easy reading
names(top_state_stats) <- categories

# --- FINAL OUTPUT ---
cat("--- Analysis for Top State:", top_state_name, "---\n")
print(top_state_stats)
```

```r
library(readr)
library(dplyr)
library(tidyr)

# 1. Load the data
titanic_df <- read_csv("titanic.csv")

# 2. Create a function to calculate survival rates and percentages
# This avoids repetitive code for Sex, Pclass, and Embarked
get_survival_stats <- function(df, column_name) {
  df %>%
    filter(!is.na(!!sym(column_name))) %>%      # Exclude NAs
    group_by(!!sym(column_name)) %>%
    summarise(
      Total_Passengers = n(),
      Survived_Count = sum(Survived == 1),
      Survival_Rate = (Survived_Count / Total_Passengers) * 100
    ) %>%
    arrange(desc(Survival_Rate))
}

# 3. Apply the function to main categories
categories <- c("Sex", "Pclass", "Embarked")
titanic_results <- lapply(categories, function(cat) get_survival_stats(titanic_df, cat))
names(titanic_results) <- categories

# 4. Specific Insight: Pclass distribution for survivors from Cherbourg (C)
cherbourg_survivors_class <- titanic_df %>%
  filter(Embarked == "C", Survived == 1) %>%
  count(Pclass) %>%
  mutate(Percentage = (n / sum(n)) * 100)

# --- PRINT RESULTS ---
print("--- Survival Statistics by Category ---")
print(titanic_results)

print("--- Class Distribution of Survivors from Cherbourg ---")
print(cherbourg_survivors_class)
```
