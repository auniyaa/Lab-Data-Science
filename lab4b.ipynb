```r
# Create a vector as input
age <- c(55,57,56,52,51,59,58,53,59,55,60,60,60,60,52,55,56,51,60,52,54,
         56,52,57,54,56,58,53,53,50,55,51,57,60,57,55,51,50,57,58)

# Convert to factor
factor_age <- factor(age)

levels(factor_age)
length(factor_age)
table(factor_age)

# Divide into 5 ranges (auto)
age_range <- cut(age,
                 breaks = 5,
                 include.lowest = TRUE,
                 right = TRUE)
# Frequency of each range
table(age_range)
```

```r
# Vector V1
V1 <- c(2,3,1,5,4,6,8,7,9)

# 1. Create 3x3 Matrix-1
Matrix1 <- matrix(V1, nrow = 3, ncol = 3, byrow = TRUE)
Matrix1

# 2. Transpose Matrix-1 - Matrix-2
Matrix2 <- t(Matrix1)
Matrix2

# 3. Rename rows and columns
rownames(Matrix1) <- paste("R", 1:3, sep = "")
colnames(Matrix1) <- paste("C", 1:3, sep = "")

rownames(Matrix2) <- paste("R", 1:3, sep = "")
colnames(Matrix2) <- paste("C", 1:3, sep = "")

Matrix1
Matrix2

# 4. Add
Add <- Matrix1 + Matrix2
Add

# 5. Subtract
Subtract <- Matrix1 - Matrix2
Subtract

# 6. Multiply
Multiply <- Matrix1 * Matrix2
Multiply

# 7. Divide
Divide <- Matrix1 / Matrix2
Divide
```

```r
# first array
Array1 <- array(1:24, dim = c(2, 4, 3))

# Print Array1
cat("Array1:\n")
for(i in 1:dim(Array1)[3]){
  cat("\nMatrix", i, ":\n")
  print(Array1[ , , i])
}

# second array
Array2 <- array(25:54, dim = c(3, 2, 5))

# rint Array2 nicely
cat("\nArray2:\n")
for(i in 1:dim(Array2)[3]){
  cat("\nMatrix", i, ":\n")
  print(Array2[ , , i])
}

#Print
second_row_Array1 <- Array1[2, , 2]
cat("\nThe second row of the second matrix of the array:\n")
print(second_row_Array1)

# Print
element_Array2 <- Array2[3, 2, 1]
cat("\nThe element in the 3rd row and 2nd column of the 1st matrix:\n")
print(element_Array2)
```
