```r
# Function to check leap year
isLeapYear <- function(year) {
  if ((year %% 4 == 0 && year %% 100 != 0) || (year %% 400 == 0)) {
    return(TRUE)
  } else {
    return(FALSE)
  }
}

# Input
year <- as.integer(readline(prompt="Enter year: "))

# Check and print result
if (isLeapYear(year)) {
  cat(year, "is a leap year.\n")
} else {
  cat(year, "is not a leap year.\n")
}
```

```r
# Input
n <- as.integer(readline(prompt="Input an integer: "))

# Loop to display cube of each number
for (i in 1:n) {
  cat("Number is:", i, "and cube of the", i, "is:", i^3, "\n")
}
```

```r
# Input
num <- as.integer(readline(prompt="Input an integer: "))
num_copy <- num
sum <- 0

# Count digits
n <- nchar(as.character(num))

# Calculate sum of digits raised to power n
while (num_copy > 0) {
  digit <- num_copy %% 10
  sum <- sum + digit^n
  num_copy <- num_copy %/% 10
}

# Check and print result
if (sum == num) {
  cat(num, "is an Armstrong number.\n")
} else {
  cat(num, "is not an Armstrong number.\n")
}
```
