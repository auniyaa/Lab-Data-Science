```r
stud.data <- data.frame(
  stud_id = c(1:7),
  stud_name = c("Anastasia","Dima","Michael","Matthew","Laura","Kevin","Jonas"),
  score = c(12.5,9.0,16.5,12.0,9.0,8.0,19.0),
  attempts=c(1,3,2,3,2,1,2)
)
```

```r
# Add new column
stud.data$qualify <-c("yes","no","yes","no","no","no","yes")
# new row
stud.newdata <- data.frame(
  stud_id = c(8),
  stud_name = c("Emily"),
  score = c(14.5),
  attempts = c(1),
  qualify = c("yes")
)

# Bind the two data frames.
stud.data <- rbind(stud.data,stud.newdata)
```

```r
#display structure
str(stud.data)
cat("\n")

#number of rows and columns
print(dim(stud.data))
print(ncol(stud.data))
print(nrow(stud.data))
cat("\n")

#summary
print(summary(stud.data))
```
